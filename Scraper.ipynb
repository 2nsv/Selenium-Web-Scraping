```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import time
 
options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
# options.add_argument("--headless=new")
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
 
driver = webdriver.Chrome(options=options)


driver.get('https://www.naukrigulf.com/data-engineer-jobs')
time.sleep(5)
Jobs=[]

for n in range(3):
    jobs = driver.find_elements(By.CSS_SELECTOR, '.ng-box.srp-tuple')
    for i in jobs :
       Job_Title = i.find_element(By.CLASS_NAME , 'designation-title').text
       Company_Name = i.find_element(By.CLASS_NAME , 'info-org').text
       Job_Location = i.find_element(By.CLASS_NAME, 'info-loc').text
       Required_Experience = i.find_element(By.CLASS_NAME, 'info-exp').text
       Full_Job_Description = i.find_element(By.CLASS_NAME , 'description').text
       Jobs.append({
           'Job_Title': Job_Title ,
           'Company_Name' : Company_Name ,
           'Job_Location' : Job_Location ,
           'Required_Experience' : Required_Experience ,
           'Full_Job_Description':Full_Job_Description
       })
    time.sleep(2)
    driver.find_element(By.XPATH, "//span[@class='ico bwd']").click()
    time.sleep(5)   
   

time.sleep(2)
driver.find_element(By.XPATH, "//span[@class='ico bwd']").click()
time.sleep(5)
df = pd.DataFrame(Jobs)
df.to_csv('Jobs2.csv')
```
